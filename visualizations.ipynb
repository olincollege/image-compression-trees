```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
quad_tree_results_df = pd.read_csv('quad_tree/more_iteration_counts.csv')

for image_name, group in quad_tree_results_df.groupby('image name'):
    plt.scatter(
        group['iteration count'],
        group['runtime (s)'],
        label=image_name,
        s=50
    )

plt.xlabel('Iteration Count')
plt.ylabel('Runtime (s)')
plt.legend()
plt.title('Runtime vs. Iteration Count for Quad Tree Compression')
plt.grid()
plt.show()
```

```python
for image_name, group in quad_tree_results_df.groupby('image name'):
    plt.scatter(
        group['iteration count'],
        group['error'],
        label=image_name,
        s=50
    )

plt.xlabel('Iteration Count')
plt.ylabel('Error')
plt.legend()
plt.title('Error vs. Iteration Count for Quad Tree Compression')
plt.grid()
plt.show()
```

```python
quad_tree_compression_ratio = quad_tree_results_df['image size before (bytes)'] / quad_tree_results_df['image size after (bytes)']

for image_name, group in quad_tree_results_df.groupby('image name'):
    plt.scatter(
        group['iteration count'],
        quad_tree_compression_ratio[group.index],
        label=image_name,
        s=50
    )

plt.xlabel('Iteration Count')
plt.ylabel('Compression Ratio')
plt.legend()
plt.title('Compression Ratio vs. Iteration Count for Quad Tree Compression')
plt.grid()
plt.show()
```

```python
for image_name, group in quad_tree_results_df.groupby('image name'):
    plt.scatter(
        group['number of colors'],
        quad_tree_compression_ratio[group.index],
        label=image_name,
        s=50
    )

plt.xlabel('Number of Colors')
plt.ylabel('Compression Ratio')
plt.legend()
plt.title('Compression Ratio vs. Number of Colors for Quad Tree Compression')
plt.grid()
plt.show()
```

```python
huffman_benchmarks_df = pd.read_csv('data/huffman_benchmarks.csv')
huffman_color_benchmarks_df = huffman_benchmarks_df.iloc[1:1000]
huffman_size_benchmarks_df = huffman_benchmarks_df.iloc[1000:]

compression_ratios_colors = huffman_color_benchmarks_df['BeforeSize'] / huffman_color_benchmarks_df['AfterSize']
compression_ratios_size = huffman_size_benchmarks_df['BeforeSize'] / huffman_size_benchmarks_df['AfterSize']

plt.scatter(huffman_color_benchmarks_df['Colors'], compression_ratios_colors)
plt.xlabel('Number of Colors')
plt.ylabel('Compression Ratio')
plt.title('Compression Ratio vs. Number of Colors for Huffman Compression')
plt.grid()
plt.show()
```

```python
plt.scatter(huffman_color_benchmarks_df['Colors'], huffman_color_benchmarks_df['Runtime'])
plt.xlabel('Number of Colors')
plt.ylabel('Runtime (s)')
plt.title('Runtime vs. Number of Colors for Huffman Compression')
plt.grid()
plt.show()
```

```python
plt.scatter(huffman_size_benchmarks_df['Pixels'], huffman_size_benchmarks_df['Runtime'])
plt.xlabel('Number of Pixels')
plt.ylabel('Runtime (s)')
plt.title('Runtime vs. Number of Pixels for Huffman Compression')
plt.grid()
plt.show()
```

```python
plt.scatter(huffman_size_benchmarks_df['Pixels'], compression_ratios_size)
plt.xlabel('Number of Pixels')
plt.ylabel('Compression Ratio')
plt.title('Compression Ratio vs. Number of Pixels for Huffman Compression')
plt.grid()
plt.show()
```
